```python
import pandas as pd

df = pd.read_excel('/content/drive/MyDrive/Semester5/Statistika/aima.xlsx')

print("Data berhasil dimuat.")
print("Jumlah baris dan kolom: ", df.shape)
```

```python
df.head()
```

```python
# Tabel Frekuensi
datafrq = pd.crosstab(
    index=df["Grade"],
    columns="Frekuensi"
)

print(datafrq)
```

```python
datafrq.plot(color='yellow')

plt.show()
```

```python
datafrq.plot(
    kind='bar',
    color='yellow',
    title='Bar Chart Frekuensi Grade'
)

plt.ylabel("Frekuensi")
plt.xlabel("Grade")
plt.show()
```

```python
from matplotlib import colors

# Pie Chart
datafrq.plot(
    kind='pie',
    y='Frekuensi',
    autopct='%1.0f%%',
    legend=False,
    colors=['#FFD700', '#2E8B57', '#E74C3C', '#F1C40F', '#27AE60', '#C0392B']
)

plt.ylabel("")
plt.title("Pie Chart Frekuensi Grade")
plt.show()
```

```python
nilai = df["Final Score"]

tabel_statistik = pd.DataFrame({
    "Statistik": [
        "Mean", "Median", "Modus", "Minimum", "Maximum",
        "Q1", "Q2", "Q3", "Variance",
        "Standard Deviation", "Skewness", "Kurtosis", "Standard Error"
    ],
    "Nilai": [
        nilai.mean(),
        nilai.median(),
        nilai.mode()[0],
        nilai.min(),
        nilai.max(),
        nilai.quantile(0.25),
        nilai.quantile(0.50),
        nilai.quantile(0.75),
        nilai.var(),
        nilai.std(),
        nilai.skew(),
        nilai.kurt(),
        nilai.sem()
    ]
})

tabel_statistik.style.set_table_styles([
    {"selector": "th", "props": [("background-color", "#2E7D32"), ("color", "white")]},
    {"selector": "td", "props": [("background-color", "#FFFDE7"), ("color", "black")]}
])
```
